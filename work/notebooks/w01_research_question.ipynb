## Research Question

### Decision

Which webpages should be reviewed first?

### Unit of Analysis

One webpage (content page).

### Output

A ranked list of webpages with a refresh priority score and reason codes.

### Action

The SEO or content team reviews the highest-ranked pages and decides whether to refresh, expand, protect, prune, or monitor them.

### Cost of a Wrong Recommendation

False Positive:
A healthy page is reviewed unnecessarily, wasting time and resources.

False Negative:
A genuinely declining page is missed, potentially resulting in lost search traffic.

### Why Machine Learning?

The dataset contains thousands of pages, making manual review inefficient. Machine learning can learn patterns from historical search and engagement signals to prioritise review candidates more effectively than fixed rules alone.
